```python
df_clean = df.copy()
df_clean['date'] = pd.to_datetime(df_clean['order_date']).dt.date.astype('datetime64[ns]')
df_clean['amount'] = df_clean['quantity'] * df_clean['price']
```

```python
df_user = df_clean.groupby('customer_id', as_index=False) \
    .agg(order_cnt=('order_id', 'nunique'),
        max_order_date=('date', 'max'),
        total_order_value=('amount', 'sum'))
```

```python
today = df_clean['date'].max()
df_user['day_since_last_order'] = (today - df_user['max_order_date']).dt.days
```

```python
import numpy as np

df_user['recency_score'] = pd.cut(df_user['day_since_last_order'],
    bins=[df_user['day_since_last_order'].min(),
        np.percentile(df_user['day_since_last_order'], 20),
        np.percentile(df_user['day_since_last_order'], 40),
        np.percentile(df_user['day_since_last_order'], 60),
        np.percentile(df_user['day_since_last_order'], 80),
        df_user['day_since_last_order'].max()],
    labels=[5, 4, 3, 2, 1],
    include_lowest=True).astype(int)
```

```python
# Menghitung batas bins yang unik terlebih dahulu untuk menghindari duplikasi
bins = [
    df_user['order_cnt'].min(),
    np.percentile(df_user['order_cnt'], 20),
    np.percentile(df_user['order_cnt'], 40),
    np.percentile(df_user['order_cnt'], 60),
    np.percentile(df_user['order_cnt'], 80),
    df_user['order_cnt'].max()
]
unique_bins = sorted(list(set(bins)))

# Menyesuaikan label berdasarkan jumlah interval (jumlah bin unik - 1)
labels = list(range(1, len(unique_bins)))

df_user['frequency_score'] = pd.cut(df_user['order_cnt'],
    bins=unique_bins,
    labels=labels,
    include_lowest=True).astype(int)

df_user
```

```python
df_user['monetary_score'] = pd.cut(df_user['total_order_value'],
    bins=[df_user['total_order_value'].min(),
        np.percentile(df_user['total_order_value'], 20),
        np.percentile(df_user['total_order_value'], 40),
        np.percentile(df_user['total_order_value'], 60),
        np.percentile(df_user['total_order_value'], 80),
        df_user['total_order_value'].max()],
    labels=[1, 2, 3, 4, 5],
    include_lowest=True).astype(int)
```

```python
df_user['segment'] = np.select(
    [(df_user['recency_score'] == 5) & (df_user['frequency_score'] >= 4),
     (df_user['recency_score'].between(3, 4)) & (df_user['frequency_score'] >= 4),
     (df_user['recency_score'] >= 4) & (df_user['frequency_score'].between(2, 3)),
     (df_user['recency_score'] <= 2) & (df_user['frequency_score'] == 5),
     (df_user['recency_score'] == 3) & (df_user['frequency_score'] == 3),
     (df_user['recency_score'] == 5) & (df_user['frequency_score'] == 1),
     (df_user['recency_score'] == 4) & (df_user['frequency_score'] == 1),
     (df_user['recency_score'] <= 2) & (df_user['frequency_score'].between(3, 4)),
     (df_user['recency_score'] == 3) & (df_user['frequency_score'] <= 2),
     (df_user['recency_score'] <= 2) & (df_user['frequency_score'] <= 2)],
    ['01-Champion', '02-Loyal Customers', '03-Potential Loyalists', '04-Can\'t Lose Them',
     '05-Need Attention', '06-New Customers', '07-Promising', '08-At Risk', '09-About to Sleep',
     '10-Hibernating'],
    default='Others'
)

df_user
```

```python
pd.pivot_table(df_user, index='segment',
    values=['customer_id', 'day_since_last_order', 'order_cnt', 'total_order_value'],
    aggfunc={'customer_id': pd.Series.nunique,
             'day_since_last_order': [np.mean, np.median],
             'order_cnt': [np.mean, np.median],
             'total_order_value': [np.mean, np.median]})
```

```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules

import warnings
warnings.filterwarnings('ignore')

products = [
    'red hanging heart t-light holder', 'white hanging heart t-light holder',
    'sweetheart ceramic trinket box', 'strawberry ceramic trinket box',
    'toilet metal sign', 'bathroom metal sign',
    'red retrospot sugar jam bowl', 'red retrospot small milk jug',
    'painted metal pears assorted', 'assorted colour bird ornament',
    'regency cakestand 3 tier', 'pack of 72 retro spot cake cases',
    'jumbo bag red retrospot', 'red retrospot mug'
]

strong_pairs = [
    ('red hanging heart t-light holder', 'white hanging heart t-light holder'),
    ('sweetheart ceramic trinket box', 'strawberry ceramic trinket box'),
    ('toilet metal sign', 'bathroom metal sign'),
    ('red retrospot sugar jam bowl', 'red retrospot small milk jug'),
    ('painted metal pears assorted', 'assorted colour bird ornament'),
    ('regency cakestand 3 tier', 'pack of 72 retro spot cake cases'),
    ('jumbo bag red retrospot', 'red retrospot mug')
]

np.random.seed(42)
n_transactions = 3000
data = []

for i in range(1, n_transactions + 1):
    prefix = 'C' if i % 40 == 0 else ''
    order_id = f"{prefix}5{i:05d}"

    pair = strong_pairs[i % len(strong_pairs)]
    items = list(pair)

    if i % 2 == 0:
        extra = np.random.choice([p for p in products if p not in items], size=1, replace=False)
        items.extend(extra)

    for prod in items:
        data.append({
            'order_id': order_id,
            'product_code': str(np.random.randint(10000, 99999)),
            'product_name': prod,
            'quantity': np.random.randint(1, 10),
            'order_date': '2010-05-18 10:05:00',
            'price': 2.55,
            'customer_id': 14590.0,
            'year_month': '2010-05'
        })

df = pd.DataFrame(data)

df_clean = df.copy()
df_clean = df_clean[df_clean['order_id'].astype(str).str[:1] != 'C']

basket = pd.pivot_table(df_clean, index='order_id', columns='product_name', values='product_code',
                        aggfunc='nunique', fill_value=0)

def encode(x):
    if x == 0:
        return False
    if x > 0:
        return True

basket_encode = basket.applymap(encode)

basket_filter = basket_encode[(basket_encode > 0).sum(axis=1) > 1]
frequent_itemset = apriori(basket_filter, min_support=0.01, use_colnames=True) \
    .sort_values('support', ascending=False).reset_index(drop=True)
frequent_itemset['product_cnt'] = frequent_itemset['itemsets'].apply(lambda x: len(x))

product_association = association_rules(frequent_itemset, metric='confidence', min_threshold=0.7) \
    .sort_values(['support', 'confidence'], ascending=[False, False]) \
    .reset_index(drop=True)

product_association
```
